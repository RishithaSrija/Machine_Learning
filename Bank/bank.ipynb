```python
import numpy as np
import pandas as pd

df=pd.read_csv('bank_marketing.csv')
df.head()
```

```python
df=df[['age','job','balance','loan','contact']]
df.head()
```

```python
df.isnull().sum
```

```python
x=df.drop('contact',axis=1)
y=df['contact']
x.head(),y.head()
```

```python

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

for col in x.columns:
    x[col] = le.fit_transform(x[col])
df.head()
```

```python
from sklearn.tree import DecisionTreeClassifier,plot_tree
model=DecisionTreeClassifier(criterion='gini',max_depth=4,random_state=42)
model.fit(x,y)
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
plt.figure(figsize=(20,10))
plot_tree(
    model,
    feature_names=x.columns,
   class_names=['Likely to subscribe','Not likely to subscribe'])
plt.show()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)

model.fit(X_train, y_train)
```

```python
num_cols = x.select_dtypes(include=['int64','float64']).columns

Q1 = x[num_cols].quantile(0.25)
Q3 = x[num_cols].quantile(0.75)
IQR = Q3 - Q1

mask = ~((x[num_cols] < (Q1 - 1.5 * IQR)) |
         (x[num_cols] > (Q3 + 1.5 * IQR))).any(axis=1)

X_clean = x.loc[mask]
y_clean = y.loc[mask]


# print(outliers_left)
```

```python
Q1 = X_clean[num_cols].quantile(0.25)
Q3 = X_clean[num_cols].quantile(0.75)
IQR = Q3 - Q1

print(((X_clean[num_cols] < (Q1 - 1.5 * IQR)) |
       (X_clean[num_cols] > (Q3 + 1.5 * IQR))).sum())
```

```python
import matplotlib.pyplot as plt

x[num_cols].boxplot(figsize=(12,6))
plt.show()
```

```python
from sklearn.metrics import accuracy_score

y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
```
